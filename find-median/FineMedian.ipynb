````python
# MedianFinder Algorithm Explanation

## Problem Summary
We need to find the median of a stream of integers efficiently. The median is:
- For odd-length arrays: the middle element
- For even-length arrays: the average of the two middle elements

## Solution Approach: Two Heaps

### Key Insight
We can maintain two heaps to keep track of the smaller and larger halves of our numbers:

1. **Max Heap** (`max_heap`): Stores the smaller half of numbers
2. **Min Heap** (`min_heap`): Stores the larger half of numbers

### Heap Properties
- `max_heap` contains the smaller half (≤ median)
- `min_heap` contains the larger half (≥ median)
- Size difference: `len(max_heap) - len(min_heap)` ∈ {0, 1}

### Algorithm Steps

#### Adding a Number (`addNum`):
1. Add the number to `max_heap` (using negative values for max heap behavior)
2. Move the largest element from `max_heap` to `min_heap` 
3. If `min_heap` becomes larger than `max_heap`, rebalance by moving one element back

#### Finding Median (`findMedian`):
- If `max_heap` has more elements: median = top of `max_heap`
- If equal sizes: median = average of both heap tops

## Complexity Analysis

### Time Complexity:
- `addNum(num)`: **O(log n)** - heap operations
- `findMedian()`: **O(1)** - just accessing heap tops

### Space Complexity:
- **O(n)** - storing all n numbers in the heaps

## Why This Works
- The two-heap approach ensures we always know where the median is
- Heap operations maintain the sorted property efficiently
- Balancing keeps the median accessible in O(1) time

## Alternative Approaches
1. **Sorted Array**: O(n) insert, O(1) median → too slow for insertions
2. **Binary Search + Insert**: O(n) insert, O(1) median → still too slow
3. **Two Heaps**: O(log n) insert, O(1) median → optimal for this problem
````

````python
# MedianFinder Implementation from Scratch (No heapq library)

class MinHeap:
    """Custom Min Heap implementation"""
    
    def __init__(self):
        self.heap = []
    
    def push(self, val):
        """Add a value to the heap"""
        self.heap.append(val)
        self._heapify_up(len(self.heap) - 1)
    
    def pop(self):
        """Remove and return the minimum value"""
        if not self.heap:
            raise IndexError("pop from empty heap")
        
        # Replace root with last element
        min_val = self.heap[0]
        self.heap[0] = self.heap[-1]
        self.heap.pop()
        
        # Restore heap property
        if self.heap:
            self._heapify_down(0)
        
        return min_val
    
    def peek(self):
        """Return the minimum value without removing it"""
        if not self.heap:
            raise IndexError("peek from empty heap")
        return self.heap[0]
    
    def size(self):
        """Return the number of elements in the heap"""
        return len(self.heap)
    
    def _heapify_up(self, index):
        """Move element up to maintain heap property"""
        while index > 0:
            parent_index = (index - 1) // 2
            if self.heap[index] >= self.heap[parent_index]:
                break
            # Swap with parent
            self.heap[index], self.heap[parent_index] = self.heap[parent_index], self.heap[index]
            index = parent_index
    
    def _heapify_down(self, index):
        """Move element down to maintain heap property"""
        while True:
            left_child = 2 * index + 1
            right_child = 2 * index + 2
            smallest = index
            
            # Find the smallest among parent and children
            if left_child < len(self.heap) and self.heap[left_child] < self.heap[smallest]:
                smallest = left_child
            
            if right_child < len(self.heap) and self.heap[right_child] < self.heap[smallest]:
                smallest = right_child
            
            # If heap property is satisfied, stop
            if smallest == index:
                break
            
            # Swap and continue
            self.heap[index], self.heap[smallest] = self.heap[smallest], self.heap[index]
            index = smallest


class MaxHeap:
    """Custom Max Heap implementation (using MinHeap with negated values)"""
    
    def __init__(self):
        self.min_heap = MinHeap()
    
    def push(self, val):
        """Add a value to the max heap"""
        self.min_heap.push(-val)  # Negate for max heap behavior
    
    def pop(self):
        """Remove and return the maximum value"""
        return -self.min_heap.pop()  # Negate back to original value
    
    def peek(self):
        """Return the maximum value without removing it"""
        return -self.min_heap.peek()  # Negate back to original value
    
    def size(self):
        """Return the number of elements in the heap"""
        return self.min_heap.size()


print("Custom heap implementations created successfully!")
print("Testing MinHeap:")
min_heap = MinHeap()
for val in [5, 2, 8, 1, 9]:
    min_heap.push(val)
    print(f"Pushed {val}, min = {min_heap.peek()}")

print("\nTesting MaxHeap:")
max_heap = MaxHeap()
for val in [5, 2, 8, 1, 9]:
    max_heap.push(val)
    print(f"Pushed {val}, max = {max_heap.peek()}")
````

# C Implementation of MedianFinder

Now let's implement the same MedianFinder solution in C using the two-heap approach. In C, we need to implement our own heap data structures since there's no built-in heap library.

## Key Challenges in C:
1. **Manual Memory Management**: Need to handle malloc/free
2. **No Built-in Dynamic Arrays**: Need to implement resizable arrays
3. **No Built-in Heaps**: Must implement heap operations from scratch
4. **Pointer Management**: Careful handling of struct pointers

## Strategy:
- Use dynamic arrays to store heap elements
- Implement heapify operations manually
- Use same two-heap algorithm as Python version

````python
#include <stdio.h>
#include <stdlib.h>
#include <stdbool.h>

// Heap structure for dynamic array-based heap
typedef struct {
    int* data;          // Dynamic array to store heap elements
    int size;           // Current number of elements
    int capacity;       // Current capacity of the array
    bool is_max_heap;   // true for max heap, false for min heap
} Heap;

// MedianFinder structure containing two heaps
typedef struct {
    Heap* max_heap;     // Stores smaller half (max heap)
    Heap* min_heap;     // Stores larger half (min heap)
} MedianFinder;

// Helper function to create a heap
Heap* createHeap(bool is_max_heap) {
    Heap* heap = (Heap*)malloc(sizeof(Heap));
    heap->data = (int*)malloc(4 * sizeof(int));  // Initial capacity of 4
    heap->size = 0;
    heap->capacity = 4;
    heap->is_max_heap = is_max_heap;
    return heap;
}

// Helper function to resize heap when needed
void resizeHeap(Heap* heap) {
    if (heap->size >= heap->capacity) {
        heap->capacity *= 2;
        heap->data = (int*)realloc(heap->data, heap->capacity * sizeof(int));
    }
}

// Helper function to swap two elements
void swap(int* a, int* b) {
    int temp = *a;
    *a = *b;
    *b = temp;
}

// Heapify up operation (used after insertion)
void heapifyUp(Heap* heap, int index) {
    if (index == 0) return;  // Reached root
    
    int parent = (index - 1) / 2;
    bool should_swap = false;
    
    if (heap->is_max_heap) {
        // Max heap: parent should be >= child
        should_swap = heap->data[parent] < heap->data[index];
    } else {
        // Min heap: parent should be <= child
        should_swap = heap->data[parent] > heap->data[index];
    }
    
    if (should_swap) {
        swap(&heap->data[parent], &heap->data[index]);
        heapifyUp(heap, parent);
    }
}

// Heapify down operation (used after deletion)
void heapifyDown(Heap* heap, int index) {
    int left = 2 * index + 1;
    int right = 2 * index + 2;
    int target = index;
    
    // Find the target index based on heap type
    if (heap->is_max_heap) {
        // Max heap: find the largest among parent and children
        if (left < heap->size && heap->data[left] > heap->data[target]) {
            target = left;
        }
        if (right < heap->size && heap->data[right] > heap->data[target]) {
            target = right;
        }
    } else {
        // Min heap: find the smallest among parent and children
        if (left < heap->size && heap->data[left] < heap->data[target]) {
            target = left;
        }
        if (right < heap->size && heap->data[right] < heap->data[target]) {
            target = right;
        }
    }
    
    // If target changed, swap and continue heapifying
    if (target != index) {
        swap(&heap->data[index], &heap->data[target]);
        heapifyDown(heap, target);
    }
}

// Push element to heap
void heapPush(Heap* heap, int val) {
    resizeHeap(heap);  // Ensure capacity
    heap->data[heap->size] = val;
    heapifyUp(heap, heap->size);
    heap->size++;
}

// Pop element from heap (returns the root)
int heapPop(Heap* heap) {
    if (heap->size == 0) {
        return -1;  // Error case - should not happen in our usage
    }
    
    int root = heap->data[0];
    heap->data[0] = heap->data[heap->size - 1];
    heap->size--;
    
    if (heap->size > 0) {
        heapifyDown(heap, 0);
    }
    
    return root;
}

// Peek at top element without removing
int heapPeek(Heap* heap) {
    if (heap->size == 0) {
        return -1;  // Error case
    }
    return heap->data[0];
}

// Get heap size
int heapSize(Heap* heap) {
    return heap->size;
}

// Free heap memory
void freeHeap(Heap* heap) {
    if (heap) {
        free(heap->data);
        free(heap);
    }
}

// MedianFinder implementation

MedianFinder* medianFinderCreate() {
    MedianFinder* obj = (MedianFinder*)malloc(sizeof(MedianFinder));
    obj->max_heap = createHeap(true);   // Max heap for smaller half
    obj->min_heap = createHeap(false);  // Min heap for larger half
    return obj;
}

void medianFinderAddNum(MedianFinder* obj, int num) {
    // Strategy: Same as Python version
    // 1. Add to max_heap first (smaller half)
    // 2. Move largest from max_heap to min_heap
    // 3. Balance if needed
    
    // Add to max_heap first
    heapPush(obj->max_heap, num);
    
    // Move largest from max_heap to min_heap
    if (heapSize(obj->max_heap) > 0) {
        int largest_small = heapPop(obj->max_heap);
        heapPush(obj->min_heap, largest_small);
    }
    
    // Balance the heaps: max_heap should have equal or one more element
    if (heapSize(obj->min_heap) > heapSize(obj->max_heap)) {
        int smallest_large = heapPop(obj->min_heap);
        heapPush(obj->max_heap, smallest_large);
    }
}

double medianFinderFindMedian(MedianFinder* obj) {
    if (heapSize(obj->max_heap) > heapSize(obj->min_heap)) {
        // Odd number of elements, median is top of max_heap
        return (double)heapPeek(obj->max_heap);
    } else {
        // Even number of elements, median is average of both tops
        return (heapPeek(obj->max_heap) + heapPeek(obj->min_heap)) / 2.0;
    }
}

void medianFinderFree(MedianFinder* obj) {
    if (obj) {
        freeHeap(obj->max_heap);
        freeHeap(obj->min_heap);
        free(obj);
    }
}

// Test function to demonstrate usage
void testMedianFinder() {
    printf("=== Testing C MedianFinder Implementation ===\n");
    
    MedianFinder* obj = medianFinderCreate();
    
    // Test case from the problem
    medianFinderAddNum(obj, 1);
    printf("Added 1\n");
    
    medianFinderAddNum(obj, 2);
    printf("Added 2\n");
    
    double median1 = medianFinderFindMedian(obj);
    printf("findMedian() = %.1f\n", median1);  // Should be 1.5
    
    medianFinderAddNum(obj, 3);
    printf("Added 3\n");
    
    double median2 = medianFinderFindMedian(obj);
    printf("findMedian() = %.1f\n", median2);  // Should be 2.0
    
    // Additional test
    medianFinderAddNum(obj, 4);
    printf("Added 4\n");
    
    double median3 = medianFinderFindMedian(obj);
    printf("findMedian() = %.1f\n", median3);  // Should be 2.5
    
    medianFinderFree(obj);
    printf("Memory freed successfully\n");
}

// Uncomment the line below if you want to run the test
// int main() { testMedianFinder(); return 0; }
````

# 🏗️ Building a Max Heap from Scratch

## What is a Heap?

A **heap** is a complete binary tree where each parent node follows a specific relationship with its children:
- **Max Heap**: Parent ≥ Children (root contains maximum element)
- **Min Heap**: Parent ≤ Children (root contains minimum element)

## Key Properties:
1. **Complete Binary Tree**: All levels filled except possibly the last level (filled left to right)
2. **Heap Property**: Parent-child relationship maintained throughout
3. **Array Representation**: Can be stored efficiently in an array

## Array to Tree Mapping:
```
Array: [50, 30, 20, 15, 10, 8, 16]
Tree:       50
          /    \
        30      20  
       /  \    /  \
     15   10  8   16
```

**Index Relationships:**
- Parent of index `i`: `(i-1) // 2`
- Left child of index `i`: `2*i + 1` 
- Right child of index `i`: `2*i + 2`

````python
class MaxHeapFromScratch:
    """
    Max Heap implementation with step-by-step heap creation from a list
    """
    
    def __init__(self):
        self.heap = []
    
    def parent(self, i):
        """Get parent index"""
        return (i - 1) // 2
    
    def left_child(self, i):
        """Get left child index"""
        return 2 * i + 1
    
    def right_child(self, i):
        """Get right child index"""
        return 2 * i + 2
    
    def has_parent(self, i):
        """Check if node has parent"""
        return self.parent(i) >= 0
    
    def has_left_child(self, i):
        """Check if node has left child"""
        return self.left_child(i) < len(self.heap)
    
    def has_right_child(self, i):
        """Check if node has right child"""
        return self.right_child(i) < len(self.heap)
    
    def swap(self, i, j):
        """Swap two elements in the heap"""
        self.heap[i], self.heap[j] = self.heap[j], self.heap[i]
        print(f"  Swapped indices {i} and {j}: {self.heap}")
    
    def heapify_up(self, index):
        """
        Move element up the tree to maintain max heap property
        Used after insertion; index is the index of the newly added element
        """
        print(f"  Heapifying up from index {index} (value: {self.heap[index]})")
        
        while (self.has_parent(index) and 
               self.heap[self.parent(index)] < self.heap[index]):
            parent_idx = self.parent(index)
            print(f"    Parent {self.heap[parent_idx]} < Child {self.heap[index]}")
            self.swap(parent_idx, index)
            index = parent_idx
        
        print(f"  Heapify up complete. Final heap: {self.heap}")
    
    def heapify_down(self, index):
        """
        Move element down the tree to maintain max heap property
        Used after deletion
        """
        print(f"  Heapifying down from index {index} (value: {self.heap[index]})")
        
        while self.has_left_child(index):
            # Find the larger child
            larger_child_index = self.left_child(index)
            
            if (self.has_right_child(index) and 
                self.heap[self.right_child(index)] > self.heap[larger_child_index]):
                larger_child_index = self.right_child(index)
            
            # If heap property is satisfied, stop
            if self.heap[index] >= self.heap[larger_child_index]:
                break
            
            print(f"    Parent {self.heap[index]} < Child {self.heap[larger_child_index]}")
            self.swap(index, larger_child_index)
            index = larger_child_index
        
        print(f"  Heapify down complete. Final heap: {self.heap}")
    
    def insert(self, value):
        """Insert a value into the heap"""
        print(f"\n🔄 Inserting {value}")
        self.heap.append(value)
        print(f"  Added to end: {self.heap}")
        self.heapify_up(len(self.heap) - 1)
    
    def extract_max(self):
        """Remove and return the maximum element"""
        if not self.heap:
            return None
        
        print(f"\n🔄 Extracting maximum: {self.heap[0]}")
        
        # Replace root with last element
        max_val = self.heap[0]
        self.heap[0] = self.heap[-1]
        self.heap.pop()
        
        print(f"  Replaced root with last element: {self.heap}")
        
        # Heapify down to restore heap property
        if self.heap:
            self.heapify_down(0)
        
        return max_val
    
    def peek(self):
        """Return the maximum element without removing it"""
        return self.heap[0] if self.heap else None
    
    def size(self):
        """Return the number of elements"""
        return len(self.heap)
    
    def display_tree(self):
        """Display the heap as a tree structure"""
        if not self.heap:
            print("Empty heap")
            return
            
        print(f"\n🌳 Tree representation:")
        self._print_tree(0, 0)
    
    def _print_tree(self, index, level):
        """Helper method to print tree structure"""
        if index >= len(self.heap):
            return
        
        # Print right subtree first (appears on top)
        if self.has_right_child(index):
            self._print_tree(self.right_child(index), level + 1)
        
        # Print current node
        print("  " * level + f"{self.heap[index]}")
        
        # Print left subtree
        if self.has_left_child(index):
            self._print_tree(self.left_child(index), level + 1)

print("MaxHeapFromScratch class created! 🎉")
````

````python
def create_max_heap_from_list(numbers):
    """
    Method 1: Build heap by inserting elements one by one
    Time Complexity: O(n log n)
    """
    print("=" * 60)
    print(f"🎯 CREATING MAX HEAP FROM LIST: {numbers}")
    print("=" * 60)
    print("Method 1: Insert elements one by one")
    
    heap = MaxHeapFromScratch()
    
    for num in numbers:
        heap.insert(num)
        heap.display_tree()
        print(f"Current heap array: {heap.heap}")
    
    print(f"\n✅ Final Max Heap: {heap.heap}")
    print(f"Maximum element: {heap.peek()}")
    return heap

# Test with a sample list
test_numbers = [4, 10, 3, 5, 1, 8, 6]
max_heap = create_max_heap_from_list(test_numbers)
````

# 📝 Summary: Parent Index Formula Explanation

## Quick Answer:
**We use `(i - 1) // 2` because Python (and most programming languages) use 0-based indexing, not 1-based indexing.**

## The Complete Picture:

### **Historical Context:**
- **Mathematics/Theory**: Heaps are often taught with 1-based indexing where `parent = i // 2`
- **Programming Reality**: Arrays start at index 0, so we need `parent = (i - 1) // 2`

### **The Pattern:**
```
1-based (theoretical):  [X, 50, 30, 20, 15, 10, 8, 16]
                         0   1   2   3   4   5  6   7
                        
0-based (Python):       [50, 30, 20, 15, 10, 8, 16]
                         0   1   2   3   4   5   6
```

### **Why the "-1" is Essential:**
- **Without -1**: `i // 2` creates wrong parent-child relationships
- **With -1**: `(i - 1) // 2` correctly maps 0-based indices to tree structure
- **The subtraction compensates** for starting at 0 instead of 1

### **Memory Trick:**
Think of it as "**converting from 1-based to 0-based**":
1. **Conceptually** think in 1-based: parent would be `(i+1) // 2`  
2. **Convert to 0-based**: subtract 1 from both i and result: `((i+1) - 1) // 2 - 1 = (i - 1) // 2`

### **Real-World Impact:**
Using the wrong formula would completely break heap operations, causing:
- ❌ Incorrect parent-child relationships
- ❌ Heap property violations  
- ❌ Wrong heapify operations
- ❌ Incorrect heap behavior

**Bottom line**: The `(i - 1) // 2` formula is mathematically derived to work perfectly with 0-based array indexing!

# 🔄 Why Doesn't the Swap Function Return Anything?

## The Short Answer:
**The `swap` function modifies the heap array in-place, so no return value is needed!**

## Key Concepts:

### **1. Mutable vs Immutable Objects**
- **Lists are mutable**: Changes to list elements persist after function calls
- **Integers are immutable**: Changes to integer variables don't persist

### **2. In-Place Modification**
- `self.heap[i], self.heap[j] = self.heap[j], self.heap[i]` directly modifies the list
- The heap object maintains a reference to the same list in memory
- Changes are immediately visible to all code that accesses `self.heap`

### **3. Reference vs Value**
- `self.heap` is a reference to a list object in memory
- When we modify `self.heap[i]`, we're changing the actual list
- No need to return and reassign - the change is already "live"

````python
def demonstrate_swap_behavior():
    """
    Demonstrate why swap doesn't need to return anything
    """
    print("🔍 DEMONSTRATING SWAP BEHAVIOR")
    print("=" * 60)
    
    # Example 1: Show how list modification works in-place
    print("📝 Example 1: Basic List Modification")
    my_list = [10, 20, 30, 40, 50]
    print(f"Original list: {my_list}")
    print(f"List ID in memory: {id(my_list)}")
    
    # Swap elements directly
    my_list[1], my_list[3] = my_list[3], my_list[1]
    print(f"After swap: {my_list}")
    print(f"List ID in memory: {id(my_list)} (same object!)")
    
    print()
    
    # Example 2: Function that modifies list in-place
    print("📝 Example 2: Function with In-Place Modification")
    
    def swap_no_return(arr, i, j):
        """Swap without returning - modifies original list"""
        print(f"  Before swap: arr[{i}]={arr[i]}, arr[{j}]={arr[j]}")
        arr[i], arr[j] = arr[j], arr[i]
        print(f"  After swap:  arr[{i}]={arr[i]}, arr[{j}]={arr[j]}")
        # Notice: NO return statement!
    
    test_list = [100, 200, 300, 400]
    print(f"Before function call: {test_list}")
    
    swap_no_return(test_list, 0, 2)  # No assignment needed!
    
    print(f"After function call:  {test_list}")
    print("✅ The list was modified even though function returned nothing!")
    
    print()
    
    # Example 3: What happens with immutable types
    print("📝 Example 3: Why This Doesn't Work with Immutable Types")
    
    def try_swap_integers(a, b):
        """This WON'T work - integers are immutable"""
        print(f"  Inside function: a={a}, b={b}")
        a, b = b, a  # This only affects local variables
        print(f"  After 'swap': a={a}, b={b}")
        # These changes don't affect the original variables!
    
    x, y = 10, 20
    print(f"Before function: x={x}, y={y}")
    try_swap_integers(x, y)
    print(f"After function:  x={x}, y={y}")
    print("❌ No change! Integers are immutable - changes don't persist")
    
    print()
    
    # Example 4: Demonstrate with our MaxHeap class
    print("📝 Example 4: Heap Swap in Action")
    
    class SimpleHeap:
        def __init__(self):
            self.heap = [50, 30, 20, 15, 10]
        
        def swap(self, i, j):
            """Our heap swap function"""
            print(f"    Swapping indices {i} and {j}")
            print(f"    Before: heap[{i}]={self.heap[i]}, heap[{j}]={self.heap[j]}")
            self.heap[i], self.heap[j] = self.heap[j], self.heap[i]
            print(f"    After:  heap[{i}]={self.heap[i]}, heap[{j}]={self.heap[j]}")
    
    heap = SimpleHeap()
    print(f"Original heap: {heap.heap}")
    
    heap.swap(0, 4)  # Swap first and last elements
    
    print(f"Final heap:    {heap.heap}")
    print("✅ Heap was modified in-place!")

demonstrate_swap_behavior()
````

````python
def compare_swap_approaches():
    """
    Compare different ways to implement swap functions
    """
    print("\n🔄 DIFFERENT SWAP APPROACHES")
    print("=" * 60)
    
    # Approach 1: In-place modification (what we use)
    print("✅ Approach 1: In-Place Modification (Recommended)")
    print("-" * 50)
    
    class HeapInPlace:
        def __init__(self):
            self.heap = [1, 2, 3, 4, 5]
        
        def swap(self, i, j):
            """Modifies self.heap directly"""
            self.heap[i], self.heap[j] = self.heap[j], self.heap[i]
            # No return needed!
    
    heap1 = HeapInPlace()
    print(f"Before: {heap1.heap}")
    heap1.swap(0, 4)
    print(f"After:  {heap1.heap}")
    print("✅ Simple, efficient, clear intent")
    
    print()
    
    # Approach 2: Return new list (inefficient)
    print("❌ Approach 2: Return New List (Inefficient)")
    print("-" * 50)
    
    class HeapReturnNew:
        def __init__(self):
            self.heap = [1, 2, 3, 4, 5]
        
        def swap(self, i, j):
            """Returns a new list - WASTEFUL!"""
            new_heap = self.heap.copy()  # Creates unnecessary copy
            new_heap[i], new_heap[j] = new_heap[j], new_heap[i]
            return new_heap
    
    heap2 = HeapReturnNew()
    print(f"Before: {heap2.heap}")
    heap2.heap = heap2.swap(0, 4)  # Need explicit assignment
    print(f"After:  {heap2.heap}")
    print("❌ Wastes memory, requires assignment, confusing")
    
    print()
    
    # Approach 3: Return nothing but modify (our approach)
    print("✅ Approach 3: Void Function with Side Effects (Our Choice)")
    print("-" * 60)
    
    class HeapVoidFunction:
        def __init__(self):
            self.heap = [1, 2, 3, 4, 5]
        
        def swap(self, i, j):
            """Modifies self.heap, returns None"""
            self.heap[i], self.heap[j] = self.heap[j], self.heap[i]
            return None  # Explicit, but not necessary
    
    heap3 = HeapVoidFunction()
    print(f"Before: {heap3.heap}")
    result = heap3.swap(0, 4)
    print(f"After:  {heap3.heap}")
    print(f"Return value: {result}")
    print("✅ Clear that function has side effects")
    
    print()
    
    # Memory efficiency comparison
    print("📊 MEMORY EFFICIENCY COMPARISON")
    print("-" * 40)
    print("Approach 1 (In-place):     ✅ O(1) extra space")
    print("Approach 2 (Return new):   ❌ O(n) extra space") 
    print("Approach 3 (Void):         ✅ O(1) extra space")
    print()
    print("💡 Why In-Place is Better:")
    print("• No memory allocation/deallocation")
    print("• No copying overhead") 
    print("• Faster execution")
    print("• Clear that original object is modified")

compare_swap_approaches()
````

## **Core Concept: Lists are Mutable Objects**

In Python, when you pass a list to a function, you're passing a **reference** to the list object in memory, not a copy. This means:

### ✅ **What Happens:**

### **Memory Model:**

```
                ↑
After swap:  [10, 30, 20, 15, 50]  (Same address: 0x1234)
```

### **Python (Our Case):**

## **Key Takeaways:**

### **✅ Why No Return Value is Needed:**
1. **Mutable Container**: Lists can be modified in-place
2. **Reference Semantics**: `self.heap` points to the actual list object
3. **Persistent Changes**: Modifications are immediately visible
4. **Memory Efficient**: No copying or new object creation

### **🔍 When You WOULD Need Return Values:**
- **Immutable types**: strings, tuples, integers
- **Functional programming**: avoiding side effects
- **Creating new objects**: when original shouldn't be modified

### **💡 Design Philosophy:**
The swap function is designed as a **mutating operation** - its purpose is to change the heap's state, not to compute and return a new value. This is:
- **More efficient** (no copying)
- **More intuitive** (heap is obviously modified)
- **Standard practice** in data structure implementations

**Bottom Line**: When working with mutable data structures like heaps, in-place modifications are the norm because they're efficient and the intent is clear!

# 🔍 Does the Array Get Changed or Just Reassign Index?

## **Answer: The Array Itself Gets Changed!**

When we do `self.heap[i], self.heap[j] = self.heap[j], self.heap[i]`, we're **modifying the contents** of the existing array, not creating a new array or reassigning the entire array variable.

## Key Distinctions:

### **✅ What Actually Happens (Element Modification):**
- **Same array object** in memory
- **Only specific elements** at positions `i` and `j` are changed
- **Array reference** (`self.heap`) stays the same
- **Memory address** of the array doesn't change

### **❌ What Does NOT Happen (Array Reassignment):**
- No new array is created
- `self.heap` is not reassigned to point to a different array
- No memory allocation for a new array object

````python
class MedianFinder:

    def __init__(self):
        self.max_heap = [] # holds smaller elements - negative
        self.min_heap = [] # holds larger elements
        

    def addNum(self, num: int) -> None:

        # always add first one to max_heap
        heapq.heappush(self.max_heap, -num)

        # move if exceed smallest of the right side, also make sure min_heap is not empty as if in cold start.
        if self.min_heap and (-self.max_heap[0] > self.min_heap[0]):
            val = -heapq.heappop(self.max_heap)
            heapq.heappush(self.min_heap, val)
        # need to balance heaps
        if (len(self.max_heap) > len(self.min_heap) + 1):
            val = -heapq.heappop(self.max_heap)
            heapq.heappush(self.min_heap, val)
        elif len(self.min_heap) > len(self.max_heap):
            val = -heapq.heappop(self.min_heap)
            heapq.heappush(self.max_heap, val)        

    def findMedian(self) -> float:
        if (len(self.max_heap) > len(self.min_heap)):
            val = -self.max_heap[0]
        else:
            val = (-self.max_heap[0]+self.min_heap[0])/2

        return val
````

- **Median position**: index 2 (middle)

- ✅ Industry standard
