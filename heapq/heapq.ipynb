```python
import heapq
```

```python
# Create an empty list to act as a heap
heap = []
```

```python
# Push items onto the heap
heapq.heappush(heap, 10)
heapq.heappush(heap, 5)
heapq.heappush(heap, 20)
heapq.heappush(heap, 3)
print(heap)
```

```python
# The smallest item is always at index 0
print(heap[0])
```

```python
# Pop the smallest item from the heap
smallest_item = heapq.heappop(heap)
print(smallest_item)
```

```python
# Drawing stick diagram for min heap [3, 5, 20, 10]
print("Min Heap Array: [3, 5, 20, 10]")
print("Array indices:  [0, 1, 2,  3 ]")
print()
print("Stick Diagram (Tree Structure):")
print()
print("       3(0)")
print("      /    \\")
print("   5(1)    20(2)")
print("   /")
print("10(3)")
print()
print("Tree Relationships:")
print("• Root: 3 at index 0")
print("• Left child of 3: 5 at index 1")
print("• Right child of 3: 20 at index 2") 
print("• Left child of 5: 10 at index 3")
print()
print("Index Formula Verification:")
print("• Parent of index i: (i-1)//2")
print("• Left child of index i: 2*i+1")
print("• Right child of index i: 2*i+2")
print()
print("Examples:")
print(f"• Parent of 5 (index 1): ({1}-1)//2 = {(1-1)//2} → element {[3, 5, 20, 10][(1-1)//2]}")
print(f"• Parent of 20 (index 2): ({2}-1)//2 = {(2-1)//2} → element {[3, 5, 20, 10][(2-1)//2]}")
print(f"• Parent of 10 (index 3): ({3}-1)//2 = {(3-1)//2} → element {[3, 5, 20, 10][(3-1)//2]}")
print()
print("Min Heap Property Check:")
heap_array = [3, 5, 20, 10]
for i in range(len(heap_array)):
    left_child_idx = 2*i + 1
    right_child_idx = 2*i + 2
    
    if left_child_idx < len(heap_array):
        left_child = heap_array[left_child_idx]
        print(f"• {heap_array[i]} ≤ {left_child} ✓ (parent ≤ left child)")
    
    if right_child_idx < len(heap_array):
        right_child = heap_array[right_child_idx]
        print(f"• {heap_array[i]} ≤ {right_child} ✓ (parent ≤ right child)")
```

```python
# Why can't 10 be at index 2 and 20 at index 3?
print("Let's compare both arrangements:")
print()

# Original correct heap
print("CORRECT Min Heap: [3, 5, 20, 10]")
print("Array indices:    [0, 1, 2,  3 ]")
print()
print("       3(0)")
print("      /    \\")
print("   5(1)    20(2)")
print("   /")
print("10(3)")
print()

# Alternative arrangement
print("PROPOSED arrangement: [3, 5, 10, 20]")
print("Array indices:        [0, 1, 2,  3 ]")
print()
print("       3(0)")
print("      /    \\")
print("   5(1)    10(2)")
print("   /")
print("20(3)")
print()

print("Min Heap Property Check for PROPOSED [3, 5, 10, 20]:")
proposed_heap = [3, 5, 10, 20]
is_valid = True

for i in range(len(proposed_heap)):
    left_child_idx = 2*i + 1
    right_child_idx = 2*i + 2
    
    if left_child_idx < len(proposed_heap):
        left_child = proposed_heap[left_child_idx]
        if proposed_heap[i] <= left_child:
            print(f"✓ {proposed_heap[i]} ≤ {left_child} (parent ≤ left child)")
        else:
            print(f"✗ {proposed_heap[i]} > {left_child} (VIOLATES min heap property!)")
            is_valid = False
    
    if right_child_idx < len(proposed_heap):
        right_child = proposed_heap[right_child_idx]
        if proposed_heap[i] <= right_child:
            print(f"✓ {proposed_heap[i]} ≤ {right_child} (parent ≤ right child)")
        else:
            print(f"✗ {proposed_heap[i]} > {right_child} (VIOLATES min heap property!)")
            is_valid = False

print()
if is_valid:
    print("✅ The proposed arrangement [3, 5, 10, 20] IS a valid min heap!")
    print("   Both arrangements are valid min heaps with the same elements.")
else:
    print("❌ The proposed arrangement violates the min heap property!")
    
print()
print("CONCLUSION:")
print("Actually, [3, 5, 10, 20] would also be a valid min heap!")
print("The key insight: There can be MULTIPLE valid heap arrangements")
print("for the same set of elements, as long as the heap property is maintained.")
print()
print("Why we got [3, 5, 20, 10] instead:")
print("• This depends on the ORDER of insertion")
print("• Let's trace how heapq.heappush() built our heap...")
```

```python
# Step-by-step trace of how our heap [3, 5, 20, 10] was built
import heapq

print("Tracing the insertion order that created [3, 5, 20, 10]:")
print("We pushed: 10, 5, 20, 3 (in that order)")
print()

# Step by step reconstruction
heap_trace = []

print("Step 1: heappush(heap, 10)")
heapq.heappush(heap_trace, 10)
print(f"Heap: {heap_trace}")
print("Tree: 10(0)")
print()

print("Step 2: heappush(heap, 5)")
heapq.heappush(heap_trace, 5)
print(f"Heap: {heap_trace}")
print("Tree:")
print("   5(0)")
print("  /")
print("10(1)")
print("(5 bubbled up to root because 5 < 10)")
print()

print("Step 3: heappush(heap, 20)")
heapq.heappush(heap_trace, 20)
print(f"Heap: {heap_trace}")
print("Tree:")
print("   5(0)")
print("  /    \\")
print("10(1)  20(2)")
print("(20 added as right child of root, no bubbling needed)")
print()

print("Step 4: heappush(heap, 3)")
heapq.heappush(heap_trace, 3)
print(f"Heap: {heap_trace}")
print("Tree:")
print("   3(0)")
print("  /    \\")
print("5(1)   20(2)")
print("/")
print("10(3)")
print("(3 added at index 3, then bubbled up: 3→5→root)")
print()

print("FINAL RESULT: [3, 5, 20, 10]")
print()
print("Key insight: The heap structure depends on:")
print("1. ORDER of insertion (we inserted 10, 5, 20, 3)")
print("2. The heapify/bubble-up process during insertion")
print("3. Elements are added left-to-right, level-by-level")
print("4. Then they bubble up if they're smaller than their parent")
print()
print("Both [3, 5, 20, 10] and [3, 5, 10, 20] are valid min heaps!")
print("The difference is just the insertion order and heapify process.")
```
