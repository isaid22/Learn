```python
coins = [1, 2, 5]
amount = 5

dp = [0] * (amount + 1)
dp[0] = 1  # Base case: one way to make amount 0

for coin in coins: # Iterate over each coin value 
    print(f"Processing coin: {coin}")
    for x in range(coin, amount + 1):
        #print(f"  Before: dp[{x}] = {dp[x]}")
        dp[x] += dp[x - coin]
        #print(f"  After: dp[{x}] = {dp[x]}")

print(dp[amount])  # Output the number of ways to make the amount
print(dp)
```
